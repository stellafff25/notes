```python
import numpy as np
```

```python
class KrylovEigenvalueSolver:
    def __init__(self, matrix):
        self.A = np.array(matrix, dtype=float)
        self.n = self.A.shape[0]
        self.y_seq = []
        self.p_coeffs = []
        self.eigenvalues = []
        
    def solve(self):
        # 1. Задаємо початковий вектор y_0 = [1, 0, ..., 0]
        y0 = np.zeros(self.n)
        y0[0] = 1.0
        self.y_seq.append(y0)
        
        # 2. Обчислення послідовності векторів y_k = A * y_{k-1}
        for k in range(1, self.n + 1):
            self.y_seq.append(self.A @ self.y_seq[k - 1])
            
        # 3. Формування матриці системи (стовпці - вектори y у зворотному порядку)
        M = np.column_stack([self.y_seq[self.n - 1 - i] for i in range(self.n)])
        b = -self.y_seq[self.n]
        
        # 4. Знаходження коефіцієнтів характеристичного полінома (p_1, ..., p_n)
        self.p_coeffs = np.linalg.solve(M, b)
        
        # 5. Знаходження власних значень (коренів полінома)
        # Додаємо коефіцієнт 1 перед старшим степенем
        poly_coeffs = np.concatenate(([1.0], self.p_coeffs))
        
        # Оскільки матриця симетрична, всі корені мають бути дійсними
        self.eigenvalues = np.sort(np.real(np.roots(poly_coeffs)))

    def print_results(self):
        print("1) Початковий вектор y_0:")
        print(f"   {self.y_seq[0]}")
        
        print("\n2) Послідовність векторів Крилова (y_k):")
        for k in range(1, self.n + 1):
            # Форматуємо вивід векторів для кращої читабельності
            formatted_vec = " ".join(f"{x:12.6f}" for x in self.y_seq[k])
            print(f"   y_{k} = [{formatted_vec}]")
            
        print("\n3) Коефіцієнти характеристичного полінома:")
        for i, p in enumerate(self.p_coeffs, start=1):
            print(f"   p_{i} = {p:.6f}")
            
        print("\n4) Характеристичне рівняння:")
        terms = [f"λ^{self.n}"] + [f"{p:+.6f}·λ^{self.n - i}" for i, p in enumerate(self.p_coeffs, start=1)]
        equation = " ".join(terms).replace("+ -", "- ") + " = 0"
        print(f"   {equation}")
        
        print("\n5) Знайдені власні значення (корені рівняння):")
        for i, ev in enumerate(self.eigenvalues, start=1):
            print(f"   λ_{i} = {ev:.6f}")
```

```python
A_matrix = [
    [7.22, 0.85, 0.96, 1.23],
    [0.85, 4.64, 1.40, 0.15],
    [0.96, 1.40, 11.43, 2.00],
    [1.23, 0.15, 2.00, 6.10]
]
```

```python
solver = KrylovEigenvalueSolver(A_matrix)
solver.solve()
solver.print_results()
```
