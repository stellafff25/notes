```python
import time
```

### Вхідні функції

```python
def f(x):
    return 6*x**5 - 3*x**4 + x**3 + 2*x**2 - 4*x + 2
```

```python
def df(x):
    return 30*x**4 - 12*x**3 + 3*x**2 + 4*x - 4
```

### 1. Метод бісекції

```python
def bisection_method(a, b, eps):
    if f(a) * f(b) > 0:
        return None, 0
    
    iteration = 0
    while abs(b - a) > eps:
        c = (a + b) / 2.0
        iteration += 1
        print(f"Бісекція | Ітерація {iteration}: x = {c:.6f}, |a-b| = {abs(b-a):.6f}")
        
        if f(a) * f(c) <= 0:
            b = c
        else:
            a = c
            
    return (a + b) / 2.0, iteration
```

### 2. Метод хорд

```python
def secant_method(a, b, eps):
    iteration = 0
    c_prev = a
    
    while True:
        iteration += 1
        c = a - f(a) * (b - a) / (f(b) - f(a))
        print(f"Хорди    | Ітерація {iteration}: x = {c:.6f}, |f(c)| = {abs(f(c)):.6f}")
        
        if abs(c - c_prev) < eps or abs(f(c)) < eps:
            break
            
        if f(a) * f(c) <= 0:
            b = c
        else:
            a = c
        c_prev = c
        
    return c, iteration
```

### 3. Метод Ньютона (дотичних)

```python
def newton_method(x0, eps):
    x = x0
    iteration = 0
    
    while True:
        iteration += 1
        x_next = x - f(x) / df(x)
        diff = abs(x_next - x)
        print(f"Ньютон   | Ітерація {iteration}: x = {x_next:.6f}, |x_k - x_k-1| = {diff:.6f}")
        
        if diff < eps or abs(f(x_next)) < eps:
            break
        x = x_next
        
    return x_next, iteration
```

### Виконання

```python
intervals = [(-1.0, -0.5)]
eps = 1e-4

for a, b in intervals:
    print("=" * 40)
    print(f"Аналіз для проміжку [{a}, {b}]")
    print("=" * 40)
    
    # Бісекція
    start = time.time()
    root_bis, iters_bis = bisection_method(a, b, eps)
    time_bis = time.time() - start
    print(f">>> РЕЗУЛЬТАТ (Бісекція): Корінь = {root_bis:.6f}, Ітерацій = {iters_bis}, Час = {time_bis:.5f} сек\n")
    
    # Хорди
    start = time.time()
    root_sec, iters_sec = secant_method(a, b, eps)
    time_sec = time.time() - start
    print(f">>> РЕЗУЛЬТАТ (Хорди): Корінь = {root_sec:.6f}, Ітерацій = {iters_sec}, Час = {time_sec:.5f} сек\n")
    
    # Ньютон (Початкова точка 'a' через те, що f(-1)*f''(-1) > 0)
    start = time.time()
    root_newt, iters_newt = newton_method(a, eps)
    time_newt = time.time() - start
    print(f">>> РЕЗУЛЬТАТ (Ньютон): Корінь = {root_newt:.6f}, Ітерацій = {iters_newt}, Час = {time_newt:.5f} сек\n")
```
