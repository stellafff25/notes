```python
class GaussMethodSolver:
    def __init__(self, matrix_a, vector_b):
        """Ініціалізація розв'язувача, створення розширеної матриці"""
        self.n = len(matrix_a)
        # Створюємо розширену матрицю (додаємо стовпець вільних членів b)
        self.matrix = [row[:] + [b] for row, b in zip(matrix_a, vector_b)]
        # Зберігаємо копію початкової матриці для перевірки нев'язки
        self.original_matrix = [row[:] for row in self.matrix]
        self.roots = [0.0] * self.n

    def print_state(self, title):
        """Виведення матриці з 6 значущими цифрами"""
        print(f"\n=== {title} ===")
        for row in self.matrix:
            formatted_row = " | ".join(f"{val:10.6f}" for val in row)
            print(f"[{formatted_row}]")

    def forward_pass(self):
        """Прямий хід: приведення матриці до верхнього трикутного вигляду"""
        self.print_state("Початкова розширена матриця")
        
        for k in range(self.n - 1):
            for i in range(k + 1, self.n):
                if self.matrix[k][k] == 0:
                    raise ValueError("Головний елемент дорівнює нулю")
                
                # Обчислюємо множник для поточного рядка
                multiplier = self.matrix[i][k] / self.matrix[k][k]
                
                # Віднімаємо відповідні елементи
                for j in range(k, self.n + 1):
                    self.matrix[i][j] -= multiplier * self.matrix[k][j]
            
            self.print_state(f"Матриця після кроку прямого ходу №{k + 1}")

    def backward_pass(self):
        """Зворотний хід: знаходження невідомих"""
        for i in range(self.n - 1, -1, -1):
            known_sum = sum(self.matrix[i][j] * self.roots[j] for j in range(i + 1, self.n))
            self.roots[i] = (self.matrix[i][self.n] - known_sum) / self.matrix[i][i]
        
        print("\n=== Знайдені корені рівняння ===")
        for i, root in enumerate(self.roots):
            print(f"x_{i+1} = {root:.6f}")

    def calculate_residuals(self):
        """Обчислення вектора нев'язки r = |b - Ax|."""
        print("\n=== Вектор нев'язки (похибка) ===")
        for i in range(self.n):
            # Обчислюємо суму A*x для і-того рядка початкової матриці
            ax_sum = sum(self.original_matrix[i][j] * self.roots[j] for j in range(self.n))
            residual = abs(self.original_matrix[i][self.n] - ax_sum)
            # Виводимо в експоненціальному форматі, оскільки числа дуже близькі до нуля
            print(f"r_{i+1} = {residual:.6e}")
```

```python
# Матриця коефіцієнтів A
A = [
    [7.03,  1.22,  0.85,  1.135, -0.81],
    [0.98,  3.39,  1.30, -1.63,   0.57],
    [1.09, -2.46,  6.21,  2.10,   1.033],
    [1.345, 0.16,  2.10,  5.33, -12.00],
    [1.29, -1.23, -0.767, 6.00,   1.00]
]

# Вектор вільних членів b
B = [2.1, 0.84, 2.58, 11.96, -1.47]

# Створення об'єкта та запуск методів
solver = GaussMethodSolver(A, B)
solver.forward_pass()
solver.backward_pass()
solver.calculate_residuals()
```
